```python
# df.columns
# df.info()
# df.describe()
# df.head()
df['regions'] = df ["place"].str.split(",").str[-1].str.strip()
print(df['regions'].value_counts())
```

Filtering: Apply multiple boolean conditions using & and | to find earthquakes that are both greater than magnitude 6.0 and occurred in a specific region (e.g., "Japan" or “Texas”).0   

```python
filtered = df[
    (df['mag'] > 6.0) |
    (
        df['place'].str.contains('Japan') |
        df['place'].str.contains('Alaska')
    )
]
result2 = filtered[['place', 'mag']]
print(result2)
```
