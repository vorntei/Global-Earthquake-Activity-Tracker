Data I/O: You start by loading the data using pd.read_csv(). Later, you will export their filtered findings using .to_excel() or .to_json() for a web team to use.
Initial Data Inspection: Before doing anything, you must run .info() and .describe() to understand the sheer volume of data, use .head() to see the layout, and use .value_counts() to see which regions have the most frequent seismic activity.
Indexing and Selection: 
Column/Label/Integer: Use dictionary-style selection to grab the 'Location' column. Use .iloc[] to extract the top 5 largest earthquakes based on row position.
Filtering: Apply multiple boolean conditions using & and | to find earthquakes that are both greater than magnitude 6.0 and occurred in a specific region (e.g., "Japan" or “Texas”).0              

```python
import pandas as pd
df = pd.read_csv("earthquake - earthquake.csv")
# Data I/O:
filtered = df[df['mag'] > 6.0]
filtered.to_json("output.json", orient="records", indent=4)
filtered.to_excel("output.xlsx", index=False)
```

```python
#Initial Data Inspection
df.columns
df.info()
df.describe()
df.head()
df.value_counts()
```

```python
# Indexing and Selection: 
# Column/Label/Integer
location_column = df['locationSource']
print(location_column.head())
df_sorted = df.sort_values(by='mag', ascending=False)
df.iloc[0:5]
```

```python
#Filtering:
filtered = df[
    (df['mag'] > 6.0) &
    (
        df['locationSource'].str.contains('uk') |
        df['locationSource'].str.contains('us')
    )
]
print(filtered)
```
